## Classes and Objects

```python
class Rocket:
    def __init__ (self):
        self.x = 0
        self.y = 0 
    def move_up(self):
        self.y += 1

my_rocket = Rocket ()  # Create a rocket Object
print("Rocket position: ",my_rocket.y)
my_rocket.move_up()
print("Rocket position: ",my_rocket.y)
```

```python
class Rocket:
    def __init__(self,x=0,y=0):
        self.x = x
        self.y = y
    def move_up(self):
        self.y += 1
rockets = []
rockets.append(Rocket())
rockets.append(Rocket(0,10))
rockets.append(Rocket(100,0))


for i, rocket in enumerate(rockets):
    print(f"Rocket {i} position: x = {rocket.x}, y = {rocket.y}")
```

```python
months = ['Jan','Feb', 'Mar', 'Apr']
apple_pie_sales = [34, 19, 28, 45]
blueberry_pie_sales =  [18, 6, 15, 51]
```

```python
import pandas as pd
df =pd.DataFrame({
    'Months': months,
    'Apple Pie Sales': apple_pie_sales,
    'Blueberry Pie Sales': blueberry_pie_sales
}) 

df
```

```python
Apple_Sales = df['Apple Pie Sales'].sum()
Blueberry_Sales = df['Blueberry Pie Sales'].sum()
if Apple_Sales > Blueberry_Sales:
    print("Apple Pie sold best with ",Apple_Sales,"units sold")
elif Blueberry_Sales > Apple_Sales:
    print("Blueberry Pie sold best with ",Blueberry_Sales,"units sold")
else:
    print("Both sold equally")
```

```python
apple_pie_price = 12
blueberry_pie_price = 15
df['Apple Pie Revenue'] = df['Apple Pie Sales'] * apple_pie_price
df['Blueberry Pie Revenue'] = df['Blueberry Pie Sales'] * blueberry_pie_price
df
```

```python
#For the charts
import matplotlib.pyplot as plt
# Bar Chart
df.plot(x = 'Months' , y = ['Apple Pie Sales','Blueberry Pie Sales'], kind = 'bar')
plt.ylabel('Total Pies Sold')
plt.title('Pie Sales')
plt.show()
```

```python
sales = df[['Apple Pie Sales', 'Blueberry Pie Sales']].sum()
sales.plot(labels=['Apple Pie', 'Blueberry Pie'],autopct='%1.1f%%',kind='pie')
plt.title('Total Pie Sales')
plt.show()
```

## Pluto Shorter csv

```python
import pandas as pd
df = pd.read_csv('pluto_shorter.csv' , low_memory = False)
```

```python
df.head()
```

```python
df.tail()
```

```python
df.columns
```

```python
my_cols = ['borough','numfloors','yearbuilt', 'landuse', 'zipcode', 'assesstot', 'xcoord','ycoord']
df = df[my_cols]
```

```python
df
```

```python
##IMPORTANT## Dictionary

land_use_dict = {1: 'One & Two Family Buildings',
                2: 'Multi-Family Walk-Up Buildings',
                3: 'Multi-Family Elevator Buildings',
                4: 'Mixed Residential & Commercial Buildings',
                5: 'Commercial & Office Buildings',
                6: 'Industrial & Manufacturing',
                7: 'Transportation & Utility',
                8: 'Public Facilities & Institutions',
                9: 'Open Space & Outdoor Recreation',
                10: 'Parking Facilities',
                11: 'Vacant Land'}
df['landuse'] = df['landuse'].map(land_use_dict)
```

```python
updated_names = {
    'BX':'Bronx',
    'BK':'Brooklyn',
    'MN':'Manhattan',
    'QN':'Queens',
    'SI':'Staten Island'
}
df['borough'] = df['borough'].map(updated_names)
```

```python
df
```

```python
df.head()
```

```python
# Find out the average price of a building in Manhattan (MN) compared with Queens (QN)
manhattan_avg = df[df['borough' ] == 'Manhattan']['assesstot'].mean()
queens_avg = df[df['borough'] == 'Queens']['assesstot'].mean()
print(f"Average of Manhattan: {manhattan_avg:.2f} ")
print(f"Average of Queens: {queens_avg:.2f} ")
```

```python
# What was the most expensive building built in the year you were born?
birth_year = 2003
most_expensive_building = df[df['yearbuilt'] == birth_year ]['assesstot'].max()
print(f"The most expensive building in the year {birth_year} was worth {most_expensive_building:.2f}")
```

```python
# Compare average building
average_by_borough = df['assesstot'].mean()
print(f"Average prices of buildings: {average_by_borough:.2f}")
```

```python
# BONUS: Average price for buildings in Brooklyn, Manhattan, and Queens 
# with 5-10 floors built between 2000-2010
def average_building_price(borough):
    return df[(df['borough'] == borough) & 
              (df['numfloors'].between(5,10)) & 
              (df['yearbuilt'].between(2000,2010))]['assesstot'].mean()

brooklyn_avg = average_building_price('Brooklyn')
manhattan_avg = average_building_price('Manhattan')
queens_avg = average_building_price('Queens')

print(f"Average price for buildings in Brooklyn with the mention characteristics is {brooklyn_avg:.2f}")
print(f"Average price for buildings in Manhattan with the mention characteristics is {manhattan_avg:.2f}")
print(f"Average price for buildings in Queens with the mention characteristics is {queens_avg:.2f}")
```

```python
# Count unique values in the 'landuse' column for Brooklyn only
unique_brooklyn_landuse = df[df['borough'] == 'Brooklyn']['landuse'].value_counts()
print(f"Unique brooklyn landuse {unique_brooklyn_landuse}")
```

```python
# Strategy 1: Replace NaNs with mean year_built for buildings of the same landuse
df['yearbuilt_imputed_1'] = df.groupby('landuse')['yearbuilt'].transform(lambda x: x.fillna(x.mean()))

# Strategy 2: Replace NaNs with mean year_built for buildings of the same landuse in the same borough
df['yearbuilt_imputed_2'] = df.groupby(['borough', 'landuse'])['yearbuilt'].transform(lambda x: x.fillna(x.mean()))

# Check the results
print("Original 'yearbuilt' null count:", df['yearbuilt'].isnull().sum())
print("'yearbuilt_imputed_1' null count:", df['yearbuilt_imputed_1'].isnull().sum())
print("'yearbuilt_imputed_2' null count:", df['yearbuilt_imputed_2'].isnull().sum())

# Display some rows to compare
print(df[['borough', 'landuse', 'yearbuilt', 'yearbuilt_imputed_1', 'yearbuilt_imputed_2']].head(10))

```
