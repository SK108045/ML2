# Question 1

### Data Proccessing - Lametization

```python
import pandas as pd
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
import string

# Download required NLTK data
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')

# Load dataset
file_name = 'labelled sentences(1).xlsx'
df = pd.read_excel(file_name)

# Initialize lemmatizer and stopwords
lemmatizer = WordNetLemmatizer()
stop_words = set(stopwords.words('english'))

def preprocess(text):
    text = text.lower()
    tokens = word_tokenize(text)
    tokens = [
        lemmatizer.lemmatize(word)
        for word in tokens
        if word not in stop_words and word not in string.punctuation
    ]
    return ' '.join(tokens)

# Apply preprocessing
df['Processed_Sentences'] = df['Sentences'].apply(preprocess)

# Save to CSV
df.to_csv('preprocessed_sentences.csv', index=False)

print("Done! File saved as 'preprocessed_sentences.csv'")
print(df[['Sentences', 'Processed_Sentences']].head())
```

# Question 2

### Load the Processed Data

```python
import pandas as pd

# Load the preprocessed CSV
df = pd.read_csv('preprocessed_sentences.csv')
documents = df['Processed_Sentences']

print(df.head())
```

### 1. Count Vectorization

```python
from sklearn.feature_extraction.text import CountVectorizer

count_vectorizer = CountVectorizer()
X_count = count_vectorizer.fit_transform(documents)

# Convert to DataFrame
count_df = pd.DataFrame(
    X_count.toarray(),
    columns=count_vectorizer.get_feature_names_out()
)
count_df['Label'] = df['Label'].values

# Save to CSV
count_df.to_csv('count_vectorized.csv', index=False)

print("=== COUNT VECTORIZER ===")
print(f"Shape: {X_count.shape}")
print(count_df.iloc[:, :10].head())
```

### 2.TF-IDF vectorization

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf_vectorizer = TfidfVectorizer()
X_tfidf = tfidf_vectorizer.fit_transform(documents)

# Convert to DataFrame
tfidf_df = pd.DataFrame(
    X_tfidf.toarray(),
    columns=tfidf_vectorizer.get_feature_names_out()
)
tfidf_df['Label'] = df['Label'].values

# Save to CSV
tfidf_df.to_csv('tfidf_vectorized.csv', index=False)

print("=== TF-IDF VECTORIZER ===")
print(f"Shape: {X_tfidf.shape}")
print(tfidf_df.iloc[:, :10].head())
```
